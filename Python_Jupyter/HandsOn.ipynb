```python
n = 10

if (n % 2) != 0:
    print("Weird")
elif (n % 2) == 0 and (2 >= n <= 5):
    print("Not Weird")
elif (n % 2) == 0 and (n >= 6 and n <= 20):
    print("Weird")
elif (n % 2) == 0 and (n > 20):
    print("Not Weird")
```

```python
import random

roll = random.randint(1, 6)


guess = int(input("Guess the number (between 1 and 6): "))
if guess == roll:
    print("Congratulations! You guessed correctly.")
    print(f"You rolled a {roll}.")
else:
    print(f"Sorry, the number was {roll}.")
```

```python
acronyms = []

acronyms.append('LOL')
acronyms.append('BRB')
acronyms.append('IDK')
acronyms.append('SMH')
acronyms.append('TBH')
acronyms.append('FOMO')

print("Acronyms list:", acronyms)
print(acronyms[len(acronyms)-1])  # Output: FOMO

word = 'idk'.upper()

if word in acronyms:
    print(f"{word} is an acronym.")

for value in acronyms:
    print("=====")
    print(f"|{value}|")

print("=====")
```
